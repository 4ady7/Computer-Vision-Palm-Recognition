### Data Augmentation Stage

```python
#adding a data augmentation stage
dataAugmentation = keras.Sequential(
    [
        layers.RandomFlip("horizontal"), #RandomFlip("horizontal") randomlyy flips the input image left to right
        layers.RandomRotation(0.2), #RandomRotation(0.1) randomly rotates the image by 72 degrees
        layers.RandomZoom(0.2), #RandomZoom(0.2) randomly zooms in or out of the image by 20%
    ]
)
```

```python
### Building the model
```

```python
#randomly augmented training images to be displayed
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 10))
for images, _ in train_dataset.take(1):
    for i in range(9):
        augmented_images = dataAugmentation(images)
        ax = plt.subplot(3, 3, i + 1)
        plt.imshow(augmented_images[0].numpy().astype("uint8"))
        plt.axis("off")
```

### Building the model

```python
inputs = keras.Input(shape=(180, 180, 3)) #building a model with 180 pixels height, and 180 pixels width, and 3 colors
x = dataAugmentation(inputs) #assigning the inputs of the dataAugmentation variable to the variable x 
x = layers.Rescaling(1./255)(x)
x = layers.Conv2D(filters=32, kernel_size=3, activation="relu")(x) #first layer #relu stands for recitified linearr unit, and it allows the model to learn complex shapess rather than just straight lines 
x = layers.MaxPooling2D(pool_size=2)(x)
x = layers.Conv2D(filters=32, kernel_size=3, activation="relu")(x) #second layer #relu stands for recitified linearr unit, and it allows the model to learn complex shapess rather than just straight lines 
x = layers.MaxPooling2D(pool_size=2)(x)
x = layers.Conv2D(filters=32, kernel_size=3, activation="relu")(x) #third layer #relu stands for recitified linearr unit, and it allows the model to learn complex shapess rather than just straight lines 
x = layers.MaxPooling2D(pool_size=2)(x)
x = layers.Flatten()(x) #converting the value into a 1D vector, and assigning it to the variable x
x = layers.Dense(64, activation="relu")(x) #Rectified linear unit is the activation function that handles non linearity
#x = layers.Dropout(0.5)(x) #Dropout is a regularization technique to prevent overfitting
outputs = layers.Dense(5, activation="softmax")(x) #softmax activation function becuase I am using more than 2 classes. In addition, 5 represents five classes.
model = keras.Model(inputs=inputs, outputs=outputs) #creating a model object, and assiging it to the variable model
```

```python
model.summary()
```

```python
model.compile(optimizer="rmsprop", #using the root mean square resilient back propagation as the optimizer to determine how the network will be updated
    loss="sparse_categorical_crossentropy", #for comparing the desired output and the actual output of the network, and returning the difference between them, which is the measure of the error.
    metrics=["accuracy"]) #for mmonitoring and tracking the accuracy
```

### Training the model

```python
#training the model

callbacks = [
    keras.callbacks.ModelCheckpoint( #using ModelCheckpoint callback to save the model after each epoch.
        filepath="Palm_Model.keras", #naming the file where the model will be saved
        save_best_only=True, #maintaining that the saved model is the best model corresponding to its best performing training epoch, so saving the best model, and keeping the best saved model
        monitor="val_loss"), #val_loss is the metric that determines if this model is better or worse
        # plot_metrics #adding dynamnic ploting
]
```

```python
trainingResults = model.fit( #training the model, and storing the results in the variable called trainingResults
    train_dataset, #the train_dataset is the training dataset
    epochs=75, #going through the dataset 60 times, because 60 epoches equals 60 cycles 
    validation_data=validation_dataset, #the validation data is represents by the variable validation_dataset that contain all the subfolders. Each subfolder represent a validating class.
    callbacks=callbacks)
```

```python
#model.evaluate()
```

```python
import matplotlib.pyplot as plt
accuracy = trainingResults.history["accuracy"]
val_accuracy = trainingResults.history["val_accuracy"]
loss = trainingResults.history["loss"]
val_loss = trainingResults.history["val_loss"]
epochs = range(1, len(accuracy) + 1)
plt.plot(epochs, accuracy, "bo", label="Training accuracy")
plt.plot(epochs, val_accuracy, "b", label="Validation accuracy")
plt.title("Training and validation accuracy")
plt.legend()
plt.figure()
plt.plot(epochs, loss, "bo", label="Training loss")
plt.plot(epochs, val_loss, "b", label="Validation loss")
plt.title("Training and validation loss")
plt.legend()
plt.show()
```

```python
test_model = keras.models.load_model(
    "Palm_Model.keras")
test_loss, test_acc = test_model.evaluate(test_dataset)
print(f"Test accuracy: {test_acc:.3f}")
```

```python
print(f"Test Loss: {test_loss:.3f}")
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np
import seaborn as sns

from sklearn.metrics import classification_report
y_true = np.concatenate([y for x, y in validation_dataset], axis=0)
predictions = model.predict(validation_dataset)
y_pred = np.argmax(predictions, axis=1)
class_names = validation_dataset.class_names
print(f"Classes found: {class_names}")
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10,7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('predicted label')
plt.ylabel('True label')
plt.title('Confusion Matrix')
plt.show()
print("\n Classification Report:\n")
print(classification_report(y_true, y_pred, target_names=class_names))
```

#### VGG16

```python
#defining the base of the model
vgg16_base = keras.applications.vgg16.VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(180, 180, 3))
vgg16_base.trainable = False #protecting the base of the model, which contains the pre-trained patterns, by setting its trainability status to not for training. To prevent the weights from being updated during training
```

```python
vgg16_base.summary()
```

```python
Vgg16Inputs = keras.Input(shape=(180, 180, 3))
Vgg16x = dataAugmentation(inputs) #assigning the inputs of the data_augmentation variable to the variable Vgg16x 
Vgg16x = keras.applications.vgg16.preprocess_input(Vgg16x) #Normalization
Vgg16x = vgg16_base(Vgg16x) #assigning the value to the variable Vgg16x after the image interacts with the VGG16 network
Vgg16x = layers.Flatten()(Vgg16x) #converting the value into a 1D vector, and assigning it to the variable Vgg16x
Vgg16x = layers.Dense(256, activation="relu")(Vgg16x) #Rectified linear unit is the activation function that handles non linearity
Vgg16x = layers.Dropout(0.5)(Vgg16x) #Dropout is a regularization technique to prevent overfitting
Vgg16outputs = layers.Dense(4, activation="softmax")(Vgg16x) #softmax activation function becuase I am using more than 2 classes. In addition, 5 represents five classes.
VGG16model = keras.Model(inputs=Vgg16Inputs, outputs=Vgg16outputs) #creating a model object, and assiging it to the variable model
```

```python
VGG16model.fit(
    train_dataset, #loading the images
    epochs=10, #determining the number of cycles in this case, it is 10 cycles. One cycle means that the model has looked att every single image in the data set exactly once 
    validation_data=validation_dataset, #the model corresponding to its best-performing training epoch, in terms of its performance on the validation data. The validation images are used after training to evaluate the model and to see how well it can predict data it has not memorised
    callbacks=VGG16callbacks) #overwwriting and saving the model if the score is better
```
